# 消息与提示词


## 1.1消息的内部结构
- Role：消息所属的角色和类型，如system，user，assistant
- Content:消息内容
- Metadata：（可选）元数据，存储额外消息，如：消息id，响应时间，token消耗...

## 1.2消息类型
通过role区分

1. 系统消息：系统提示词，在对话开始前设定
```
{"role": "system", "content": "你是个精通编程的软件架构师"}
```
2. 用户消息：表示用户一次输入，也可以是多模态内容
```
{"role": "user", "content": "你好啊~"}
```
3. 助手消息：代表模型回复，包括生成的文本，调用哪个工具，元数据等
```
{
    "role": "assistant",
    "content": "",
    "tool_calls": [{
        "name": "get_weather",
        "args": {"location": "北京"},
        "id": "call_00_nUD2NC9QRN5Cg1GaoIkBJQ4s"
    }]
}
```
4. 工具调用消息：工具调用结果匹配的消息类型，将结果发给模型，然后模型返回assistant消息
```
{"role": "tool", "content": "今天天气很好", "tool_call_id":"call_00_nUD2NC9QRN5Cg1GaoIkBJQ4s"}
```

## 1.6.1 对话历史管理

第 1 轮：

[system, user] → AI回复 → 保存回复

第 2 轮：

[system, user, assistant, user] → AI回复 → 保存回复

第 3 轮：

[system, user, assistant, user, assistant, user] → AI回复

## 1.6.2 对话历史优化

- 针对对话历史越来越长，消耗大量tokens和成本等问题
#### 解决方法： 保存最近前N条对话，具体
    - 总是保留system消息
    - 只保留最近N次对话，丢掉更早的历史


````python
# 定义保留最近对话轮数的函数

def keep_recent_messages(messages, max_pairs=3):
    """
    保留最近的 N 轮对话
    max_pairs: 保留的对话轮数（每轮 = user + assistant）
    """
    # 分离 system 和对话
    system_msgs = [m for m in messages if m.get("role") == "system"]
    conversation_msgs = [m for m in messages if m.get("role") != "system"]
    # 只保留最近的
    recent_msgs = conversation_msgs[-(max_pairs * 2):]
    # 返回：system + 最近对话
    return system_msgs + recent_msgs
````

````python
def keep_recent_messages(messages, max_pairs=3):
    system_msgs = [m for m in messages if m.get("role") == "system"]
    conversation_msgs = [m for m in messages if m.get("role") != "system"]
    recent_msgs = conversation_msgs[-(max_pairs * 2):]
    return system_msgs + recent_msgs


long_conversation = [
	{"role": "system", "content": "你是 Python 导师"}
]
# 第 1 轮
long_conversation.append({"role": "user", "content": "什么是列表？用一句解释"})
r1 = model.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r1.content})
# 第 2 轮
long_conversation.append({"role": "user", "content": "列表和元组有什么区别？用一句解释"})
r2 = model.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r2.content})
# 第 3 轮
long_conversation.append({"role": "user", "content": "什么是字典呢？用一句解释"})
r3 = model.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r3.content})
print(f"原始消息数: {len(long_conversation)}")
# 优化：只保留最近 2 轮
optimized = keep_recent_messages(long_conversation, max_pairs=2)
print(f"优化后消息数: {len(optimized)}")
print(f"保留的内容: system + 最近2轮对话")
# 添加新的用户问题
optimized.append({"role": "user", "content": "我第一个问题问的是什么？"})
# 使用优化后的历史
response = model.invoke(optimized)
print(f"\nAI 回复: {response.content}")
print(optimized)
````

## 1.6.3 多轮对话聊天机器人

````python
from langchain_core.messages import AIMessage,SystemMessage,HumanMessage
from dotenv import  load_dotenv
import os

load_dotenv(override=True)

MODEL_NAME="deepseek-flash"
MAX_PAIRS_HISTORY=10
EXIT_WORD="quit"

model=init_chat_model(
	model=MODEL_NAME,
	model_provider="deepseek",
	api_key=os.getenv("DEEPSEEK_API_KEY"),
	api_base=os.getenv("DEEPSEEK_BASE_URL")
)

messages=[{
	"role":"system",
	"content":"你是一名耐心友好的知识助手，可以简要回答问题"
}]

i=1 #描述对话轮数

while True:
	print("\n","="*10,f'->第{i}轮对话开始<-',"="*10,"\n")
	user_input=input("请输入：")

	if user_input.lower() == EXIT_WORD:
		print("对话已结束")
		break

	messages.append({"role":"user","content":user_input})
	memory_messages=keep_recent_messages(messages,max_pairs=MAX_PAIRS_HISTORY)

	reply_content=""

	for chunk in model.stream(memory_messages):
		if chunk.content:
			print(chunk.content,end="",flush=True)
			reply_content+=chunk.content

	print("\n","="*10,f'->第{i}轮对话结束<-',"="*10,"\n")
	i+=1
	messages.append({"role":"assistant","content":reply_content})
````

## content_blocks(跨模型供应商、标准化的多模态数据结构)
- 数据结构：它是一个 list[TypedDict] 。

- 统一格式：每个 block 都有一个 type 字段，用于区分内容类型。

- 支持类型：包括 text （文本）、 image （图片）、 audio （音频）、 video （视频）、tool_call （工具调用）以及 reasoning （推理/思维链）。

### 输入格式化
对于复杂的对话（带图片和工具的结果），建议使用content_blocks列表形式构建HumanMessage或AIMessage

````python
import base64
from langchain_openai import ChatOpenAI
from dotenv import  load_dotenv
import os

load_dotenv(override=True)

def encode_image(img_path):
    """将一张本地图片转换成 Base64 编码的 Data URI 字符串,方便在文本中嵌入图片数据"""
    with open(img_path, "rb") as img_file:
        return base64.b64encode(img_file.read()).decode("utf-8")

# 图像路径
img_path = "test.png"
# 获取图像base64编码字符串
base64_image = encode_image(img_path)

model = ChatOpenAI(
	model="qwen3.8-flash",
    api_key=os.getenv("DASHSCOPE_API_KEY"),
    base_url="https://cn-beijing.maas.qianwenaiapi.com/compatible-mode/v1",
)

response=model.invoke(
	[
		HumanMessage(
			content_blocks=[
				{'type':'text','text':'这张图里有什么'},
				{'type':'image',
				 'base64':base64_image,
				 'mime_type': 'image/png'}
            ]
        )
    ]
)

print(response.content)
````

### content 与 content_blocks 的区别

上面的用法是输入方向：把标准结构传进去。反过来读模型返回时，两者分工如下。

| | content | content_blocks |
|---|---|---|
| 本质 | 存储字段，原样保存模型返回的内容 | 派生属性（property），访问时现场计算 |
| 形态 | 字符串，或各家自定义的 dict 列表 | 统一 schema 的 list[TypedDict] |
| 覆盖范围 | 只有模型返回的正文 | 正文 + tool_calls + additional_kwargs 里的 reasoning |
| 能否赋值 | 可读写 | 只读，赋值报 AttributeError |
| 跨供应商 | 各家不同，要写 if-else | 一致，直接消费 |

一句话：content 是存下来的原文，content_blocks 是按统一格式重新翻译出来的视图。

````python
from langchain_core.messages import AIMessage

msg = AIMessage(content=[
    {"type": "text", "text": "你好"},
    {"type": "image_url", "image_url": {"url": "https://example.com/a.png"}},
])

print("content        ->", msg.content)
print("content_blocks ->", msg.content_blocks)

# content_blocks 是只读属性，赋值直接报错
try:
    msg.content_blocks = []
except AttributeError as e:
    print("赋值失败       ->", e)

# 改 content，视图立刻跟着变：说明它没有单独存一份
msg.content = "改成了纯文本"
print("改完后         ->", msg.content_blocks)
````

改完 content 之后视图立刻跟着变，说明它每次访问都重新翻译，没有缓存第二份数据。

源码在 `langchain_core/messages/base.py` 的 `content_blocks` property，内部依次调用 openai、anthropic、google_genai、bedrock_converse 几个 translator。

### content_blocks 能看到 content 里没有的东西

下面这次真实调用，正文只有两个字，content_blocks 里却多出一块 reasoning，来源是 additional_kwargs 中的 reasoning_content。

````python
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage
from langchain_deepseek import ChatDeepSeek

load_dotenv(override=True)

chat = ChatDeepSeek(model="deepseek-flash")
resp = chat.invoke([HumanMessage("只回复两个字：收到")])

print("content           ->", resp.text)
print("content_blocks    ->", resp.content_blocks)
print("additional_kwargs ->", list(resp.additional_kwargs.keys()))
````

所以 content_blocks 是从整条消息聚合出来的，不只是给 content 换个格式。

### 跨供应商归一

同一张图片，OpenAI 和 Anthropic 的原始写法不同，content_blocks 翻译成同一个结构。

````python
from langchain_core.messages import AIMessage

# 同一张图片，两家供应商的原始写法不同
openai_msg = AIMessage(content=[{"type": "image_url", "image_url": {"url": "https://example.com/a.png"}}])
anthropic_msg = AIMessage(content=[{"type": "image", "source": {"type": "url", "url": "https://example.com/a.png"}}])

print("openai 原始    ->", openai_msg.content)
print("anthropic 原始 ->", anthropic_msg.content)
print("openai 归一    ->", openai_msg.content_blocks)
print("anthropic 归一 ->", anthropic_msg.content_blocks)
````

差异只在 OpenAI 那支多带一个 id，用于内部关联，语义部分一致。

### 使用建议与两个坑

- 读返回内容、要区分文本 / 图片 / 思维链 / 工具调用 → 用 content_blocks
- 原样回传厂商或做存档 → 用 content
- 只要纯文本 → 用 message.text，它只拼接 type 为 text 的块

坑一：每次访问都重新计算，部分 translator 会现场生成 id，两次读到的 id 不同，不要拿它做去重或相等判断。

坑二：构造消息时传的 content_blocks 只是语法糖，源码里就是赋给 content 字段，不会单独存一份。

```python
if content_blocks is not None:
    super().__init__(content=content_blocks, **kwargs)
```
