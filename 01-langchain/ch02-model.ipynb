# 3.模型调用方法
- invoke()阻塞式，一次性返回结果，批处理任务，无需实时反馈的场景
- ainvoke()非阻塞式，对应的异步版本，提高系统吞吐量，IO密集型任务
- stream()流式输出，实时返回中间结果
- batch()批量处理多个高并发场景，需要同时处理大量请求
- abatch()非阻塞式，对应异步版本，IO密集

## 3.3 流式输出

### stream()
流式调用 实时返回片段 ，调用后，返回一个迭代器（iterator），可以通过循环实时处理新生成的chunk块

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import AIMessage,HumanMessage,SystemMessage
from dotenv import load_dotenv
import os

load_dotenv(override=True)

DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")

model=init_chat_model(model="deepseek:deepseek-flash",
					  temperature=2.0)

for chunk in model.stream("写一首关于南邮的诗"):
	print(chunk.text,end="",flush=True)
```

## 3.4 批量调用(与invoke相比，大幅减少网络往返开销和等待时间)
   batch()允许一次性发送一组请求，模型会在后台并行处理，然后返回所有结果的列表。

### 3.4.1 batch()
等待所有请求处理完毕 按照输入顺序返回结果列表

```python
from langchain.chat_models import  init_chat_model
from dotenv import  load_dotenv
import os

load_dotenv(override=True)

DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")

model=init_chat_model(model="deepseek:deepseek-flash",
					  api_key=DEEPSEEK_API_KEY,
					  base_url=DEEPSEEK_BASE_URL,
					  temperature=1.4)

messages=["输出1","输出2","输出3"]

responses=model.batch(messages)

for response in responses:
    print(response)
```

### 3.4.2 batch_as_completed()
    按照完成顺序接受相应，结果可能乱序，但是可以将返回的元组通过开头索引进行排序

```python
from langchain.chat_models import  init_chat_model
from dotenv import  load_dotenv
import os

load_dotenv(override=True)

DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")

model=init_chat_model(model="deepseek:deepseek-flash",
					  api_key=DEEPSEEK_API_KEY,
					  base_url=DEEPSEEK_BASE_URL,
					  temperature=1.4)

messages=["输出50字随机长文","输出1","1+1=？"]

responses=model.batch_as_completed(messages)

for response in responses:
#for response in sorted(responses):
    print(response)
```

## 美化模型输出

```python
from langchain.chat_models import  init_chat_model
from langchain_core.messages import AIMessage,HumanMessage,SystemMessage
from dotenv import  load_dotenv
import os

load_dotenv(override=True)

DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")

model=init_chat_model(model="deepseek:deepseek-flash",
					  api_key=DEEPSEEK_API_KEY,
					  base_url=DEEPSEEK_BASE_URL,
					  temperature=1.4)

messages =[SystemMessage('你是一名计算机领域的专家'),
		   HumanMessage('简洁介绍一下C++20的协程')]

response=model.invoke(messages)
messages.append(AIMessage(content=response.content))
messages.append(HumanMessage("我刚才让你介绍了什么"))
response1=model.invoke(messages)
response1.pretty_print()
```
